Soal 2. Energi Harian dan Capacity Factor PLTD (10 poin | CLO-1 | TP-1, TP-3)

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# a. Menghitung energi harian berdasarkan daya dan jam operasi
energi = daya_kw * jam

# b. Menghitung capacity factor dalam persen
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Menampilkan hasil dengan f-string
print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")

```

Di program ini, saya mulai dengan menyiapkan beberapa variabel utama di bagian atas buat nyimpen data angka seperti besar daya, jam operasi, energi aktual, sama kapasitas pembangkitnya. Setelah itu, saya bikin rumus buat ngitung energi harian dengan cara mengalikan nilai daya dengan jam operasinya. Terus, saya bikin satu rumus lagi buat nyari nilai capacity factor dengan membagi energi aktual sama hasil kali kapasitas dikali 24 jam, lalu dikali 100 biar bentuknya jadi persen. Nah, di bagian akhir, saya pakai perintah print dengan format f-string buat nampilin hasil perhitungannya ke layar, di mana khusus untuk capacity factor-nya saya kasih tambahan format :.2f supaya angka di belakang komanya dibikin rapi pas dua digit

Soal 4. Data Tekanan Uap Turbin (15 poin | CLO-1 | TP-2)

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar

# Data awal tekanan uap (bar)
tekanan = [12.4, 13.1, 12.8, 14.2, 13.5]

# a. Pembacaan ke-2 (ingat index Python mulai dari 0, jadi pembacaan ke-2 itu index 1)
pembacaan_ke2 = tekanan[1]
print(f"Pembacaan ke-2: {pembacaan_ke2} bar")

# b. Dua pembacaan terakhir pakai slicing [-2:]
dua_terakhir = tekanan[-2:]
print(f"Dua pembacaan terakhir: {dua_terakhir}")

# c. Nambahin dua data baru (13.9 dan 12.6)
tekanan.append(13.9)
tekanan.append(12.6)

# Tampilkan jumlah data sekarang pakai len()
print(f"Jumlah data setelah penambahan: {len(tekanan)}")
```

Di program ini, saya mulai dengan bikin list data awal yang isinya deretan angka tekanan uap dalam satuan bar. Terus, buat nyari nilai pembacaan kedua, saya manggil indeks [1] karena di Python urutan data itu mulainya dari angka nol, lalu saya cetak hasilnya pakai f-string. Setelah itu, saya pakai teknik slicing dengan kode [-2:] buat ngambil dan nampilin dua data tekanan paling akhir dari list. Nah, kalau mau nambahin data baru kayak angka 13.9 sama 12.6 ke bagian paling belakang list, saya pakai perintah .append(). Terakhir, saya pakai fungsi len() buat ngitung total keseluruhan isi data yang ada di dalam list setelah ketambahan data baru tadi, lalu saya cetak lagi jumlah akhirnya ke layar.

Soal 6. Status Operasi Boiler (20 poin | CLO-1 | TP-3, TP-4)

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar

# Data uji sesuai soal
data_uji = [
    (550, 10),
    (540, 14),
    (530, 13),
    (530, 11),
    (520, 13)
]

# Perulangan untuk mengecek setiap data suhu dan tekanan secara otomatis
for suhu, tekanan in data_uji:
    if suhu >= 550 or tekanan >= 14:
        status = "STATUS: TRIP"
    elif suhu > 520 and tekanan > 12:
        status = "STATUS: WARNING"
    else:
        status = "STATUS: NORMAL"

    print(f"Suhu: {suhu}, Tekanan: {tekanan} -> {status}")
```

Pada program ini, saya membuat kumpulan data uji berupa list yang berisi pasangan angka suhu dan tekanan dari soal. Supaya tidak perlu mengecek satu per satu secara manual, saya memakai perulangan for agar program bisa memeriksa seluruh data secara otomatis dari atas sampai bawah. Di dalam perulangan tersebut, saya menggunakan percabangan if-elif-else untuk menentukan status boiler. Jika suhu sudah mencapai 550 ke atas atau tekanan 14 ke atas, maka statusnya akan menjadi TRIP. Namun, jika suhu di atas 520 dan tekanan di atas 12, statusnya berubah menjadi WARNING. Jika tidak memenuhi kedua kondisi itu, maka statusnya aman atau NORMAL. Terakhir, saya menggunakan perintah print dengan format f-string agar hasil pengecekan suhu dan tekanan beserta statusnya bisa langsung tampil rapi di layar.

Soal 8. Pengosongan Tangki Bahan Bakar (25 poin | CLO-1 | TP-5)

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar

# Mulai volume tangki dari 100 liter
volume = 100

# Perulangan while berjalan selama volume masih di atas 0
while volume > 0:
    # Kurangi 20 liter setiap putaran
    volume -= 20

    # Cetak volume saat ini
    print(f"Volume: {volume} liter")

    # Cek kalau volume pas di 20 liter, keluarin warning
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah loop selesai (tangki udah 0)
print("TANGKI KOSONG - Pengosongan berhenti.")
```

Di program ini, saya bikin variabel awal buat nyimpen volume tangki sebanyak 100 liter. Terus, saya pakai perulangan while supaya proses pengurangan isinya jalan terus selama volumenya masih di atas angka 0. Di dalam perulangan itu, saya kurangi volume sebanyak 20 liter tiap putaran, terus langsung saya cetak sisa volumenya ke layar. Saya juga nyelipin kondisi if buat ngecek, kalau volume pas nyampe di angka 20 liter, program bakal otomatis ngeluarin peringatan "WARNING: Hampir kosong!". Nah, pas looping-nya berhenti total karena tangkinya udah nyampe 0, saya bikin perintah print di luar perulangan buat nampilin pesan kalau tangkinya beneran udah kosong dan prosesnya selesai.

Soal 10. Pembacaan Suhu Bearing dengan Sensor Rusak (30 poin | CLO-1 | TP-5, TP-6)

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar#
# Data suhu bearing dari sensor

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Variabel untuk menghitung data yang valid
jumlah_valid = 0

# Cek data satu-satu pakai perulangan for
for i in range(len(data)):
    titik = i + 1
    nilai = data[i]

    try:
        # Coba ubah teks jadi angka desimal (float)
        suhu = float(nilai)

        # Kalau berhasil, hitungan data valid nambah 1
        jumlah_valid += 1

        # Cek kalau suhu lebih dari 70.0
        if suhu > 70.0:
            print(f"Warning: Titik {titik} suhu bearing tinggi!")

    except ValueError:
        # Kalau gagal (ketemu "N/A"), cetak pesan terus berhenti
        print(f"Titik {titik}: data tidak valid - pemeriksaan dihentikan")
        break

# Tampilkan total data valid di akhir
print(f"Jumlah data valid: {jumlah_valid}")
```

Pada program ini, saya bikin list yang berisi data teks suhu bearing, termasuk satu data yang rusak berupa tulisan "N/A". Saya juga nyiapin variabel jumlah_valid dengan nilai nol buat ngitung total data yang berhasil diperiksa nanti. Terus, saya pakai perulangan for supaya program bisa mengecek isi list satu per satu dari awal sampai akhir. Di dalam perulangan itu, saya pakai blok try-except buat mengantisipasi error. Jadi, Python bakal mencoba mengubah data teks tersebut menjadi angka desimal menggunakan float. Kalau berhasil diubah, variabel jumlah_valid bakal otomatis bertambah satu, lalu program akan langsung mengecek apakah suhunya melebihi 70.0 derajat untuk mencetak peringatan jika suhu bearing terlalu tinggi. Nah, ketika program mendeteksi data "N/A" yang gagal diubah jadi angka, program bakal otomatis masuk ke bagian except ValueError buat mencetak pesan error kalau pemeriksaan dihentikan, lalu langsung menghentikan perulangan pakai perintah break. Terakhir, di luar perulangan, saya menampilkan total keseluruhan data valid yang berhasil lolos pemeriksaan.

```python
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 11-November-2026
# Mata Kuliah: Pemrograman Dasar

# Program Pemantauan Suhu Energi / Pembangkit
# Data suhu dari beberapa sensor (ada yang normal, tinggi, dan rusak)
suhu_sensor = [65.5, 78.2, "N/A", 88.0, 60.1]

print("Mulai Pengecekan Suhu...")

# Perulangan untuk memeriksa data satu-persatu
for i in range(len(suhu_sensor)):
    nomor = i + 1
    data = suhu_sensor[i]

    # Mencoba mengubah data teks jadi angka
    try:
        nilai_suhu = float(data)

        # Pengecekan kondisi suhu
        if nilai_suhu > 75.0:
            print("Suhu di sensor", nomor, "terlalu tinggi!")
        else:
            print("Suhu di sensor", nomor, "aman.")

    except ValueError:
        # Kalau data rusak atau "N/A"
        print("Sensor", nomor, "rusak atau tidak valid.")

print("Pengecekan selesai.")
```

Pada program ini, saya membuat sebuah list penampung data suhu sensor yang di dalamnya ada berbagai variasi nilai, termasuk angka normal, suhu tinggi, sampai data rusak berupa tulisan "N/A". Terus, saya menggunakan perulangan for supaya program bisa mengecek setiap data satu per satu dari awal sampai akhir secara otomatis tanpa perlu saya tulis ulang. Di dalam perulangan tersebut, saya memakai blok try-except buat mengantisipasi error kalau ketemu data teks. Jadi, program bakal mencoba mengubah data yang diambil menjadi angka desimal dengan fungsi float. Kalau berhasil diubah, program akan langsung masuk ke percabangan if-else untuk memeriksa apakah suhu tersebut lebih dari 75.0 derajat, lalu mencetak informasi apakah suhunya aman atau terlalu tinggi. Nah, seandainya program nemu data "N/A" yang gagal diubah jadi angka, program bakal otomatis lari ke bagian except ValueError buat mencetak pesan kalau sensor tersebut rusak atau tidak valid. Setelah seluruh isi list selesai dicek, program akan keluar dari perulangan dan mencetak pesan penutup bahwa pengecekan telah selesai.
