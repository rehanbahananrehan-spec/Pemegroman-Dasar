```python
# Jawaban Latihan 4
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]
# (a) Cetak pembacaan pertama dan terakhir
print("Pembacaan pertama dan terakhir:")
print(tegangan_harian[0])
print(tegangan_harian[-1])

# (b) Tambahkan pembacaan baru
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data
print("Jumlah total data:", len(tegangan_harian))



```

```python
# Jawaban Latihan 6

pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# 1. Cetak nama unit kedua (indeks 1)
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Ubah status salah satu unit baru ke list dengan append()
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan satu unit baru ke list dengan append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 120, "status": "Beroperasi"}
pembangkit.append(unit_baru)

# 4. Cetak seluruh data
print("Seluruh data pembangkit:")
print(pembangkit)
```
