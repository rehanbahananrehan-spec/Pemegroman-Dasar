```python
# data awal belanjaan Budi
jumlah_buku = 7
harga_buku = 18500
uang_budi = 150000
jumlah_orang = 4

# hitung total, diskon 10%, sama total bayarnya
total = jumlah_buku * harga_buku
diskon = int(total * 0.1)
bayar = total - diskon

# hitung sisa kembalian uang Budi
kembalian = uang_budi - bayar

# bagi rata bayarannya ke 4 orang + cari sisanya
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang

# cek uang Budi cukup apa nggak
uang_cukup = uang_budi >= bayar

# tampilkan hasilnya
print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
# data PLTS atap rumah Rehan
panel_rehan = 12
daya_panel = 450
jam_sinar = 5
efisiensi = 0.8
kebutuhan_rumah_rehan = 25
tarif_listrik = 1444.70

# 1. hitung energi harian PLTS rumah Rehan (kWh)
energi_harian_rehan = (panel_rehan * daya_panel * jam_sinar * efisiensi) / 1000

# 2. hitung energi bulanan & nilai rupiah hematnya
energi_bulanan_rehan = energi_harian_rehan * 30
nilai_listrik_rehan = energi_bulanan_rehan * tarif_listrik

# 3. hitung kurangnya berapa kWh sehari
kekurangan_rehan = kebutuhan_rumah_rehan - energi_harian_rehan

# 4. cek PLTS rumah Rehan udah mencukupi apa belum
mencukupi_rehan = energi_harian_rehan >= kebutuhan_rumah_rehan

# cetak hasil
print("--- Data PLTS Rumah Rehan ---")
print("Energi harian (kWh) :", round(energi_harian_rehan, 1))
print("Energi bulanan (kWh):", round(energi_bulanan_rehan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik_rehan, 1))
print("Kekurangan (kWh)    :", round(kekurangan_rehan, 1))
print("Mencukupi?          :", mencukupi_rehan)
```

```python
# inputan dari soal genset & kabel
daya = 200
jam = 8
konsumsi = 0.25
harga = 13500
tangki = 1000
arus = 20
hambatan = 0.5

# hitung energi sehari
energi = daya * jam

# cari berapa liter solar dipake sehari plus total biayanya
solar = energi * konsumsi
biaya = solar * harga

# cek tangki kuat berapa hari sama sisa solarnya
tangki_cukup = tangki // solar
sisa_solar = tangki % solar

# rumus rugi daya kabel
rugi_kabel = (arus ** 2) * hambatan

# syarat rugi tinggi: diatas 150 watt dan arus minimal 20A
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

# print hasil akhir
print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", tangki_cukup, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
