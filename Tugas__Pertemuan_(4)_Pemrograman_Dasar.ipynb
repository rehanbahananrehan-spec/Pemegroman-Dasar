```python
# 1. Fungsi Input Data
nama_pembangkit = input("Masukkan Nama Pembangkit: ")
daya_listrik = float(input("Masukkan Daya Listrik Output (kW): "))
kalor_masuk = float(input("Masukkan Kalor Bahan Bakar Masuk (kW): "))

# 2. Operasi Aritmatika
# Rumus Efisiensi (%) = (Daya Output / Kalor Masuk) * 100
efisiensi = (daya_listrik / kalor_masuk) * 100
rugi_kalor = kalor_masuk - daya_listrik

# 3. Fungsi Output Hasil Perhitungan
print("\n" + "="*40)
print(f"HASIL ANALISIS EFISIENSI: {nama_pembangkit.upper()}")
print("="*40)
print(f"Daya Listrik Output : {daya_listrik:.2f} kW")
print(f"Kalor Masuk         : {kalor_masuk:.2f} kW")
print(f"Rugi-rugi Kalor     : {rugi_kalor:.2f} kW")
print(f"Efisiensi Termal    : {efisiensi:.2f} %")
print("="*40)
```
