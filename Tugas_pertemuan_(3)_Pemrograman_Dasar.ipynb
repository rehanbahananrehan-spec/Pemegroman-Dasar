```python
# Jawaban Latihan B.1 No 1
# ============================================================================
# Nama       : Rehan
# Nim        : 4232501051
# Tanggal    : 22 September 2026
# Tujuan     : Menampilkan hasil operasi aritmatika dasar (+, -, *, /, //, %)
#              dari dua input bilangan bulat pengguna.
#=============================================================================

# Input dua bilangan bulat dari pengguna
a = int(input("Masukkan bilangan pertama: "))
b = int(input("Masukkan bilangan kedua  : "))

print("\n--- Hasil Operasi Aritmatika ---")
print(f"Hasil Penjumlahan ({a} + {b})       : {a + b}")
print(f"Hasil Pengurangan ({a} - {b})       : {a - b}")
print(f"Hasil Perkalian   ({a} * {b})       : {a * b}")
print(f"Hasil Pembagian   ({a} / {b})       : {a / b}")
print(f"Hasil Pembagian Bulat ({a} // {b})  : {a // b}")
print(f"Hasil Sisa Bagi   ({a} % {b})       : {a % b}")
```

```python
# Jawaban Latihan B.1 No 2
# ============================================================================
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 22 September 2026
# Tujuan     : Memvalidasi kelulusan nilai ujian (>= 60) menggunkan operator relasional.
# ============================================================================

# Input nilai ujian
nilai = float(input("Masukkan nilai ujian (0-100): "))

# Evaluasi pernyataan kelulusan
is_lulus = nilai >= 60

print(f"Pernyataan \"Nilai {nilai}) termasuk kategori lulus (>= 60)\" bernilai: {is_lulus}")
```

```python
# Jawaban Latihan B.1 No 3
# ============================================================================
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 22 September 2026
# Tujuan     : Menentukan kelayakan mengemudi menggunakan operator logika "and"
#              dan penerapan tanda kurung untuk operator precedence.
# ============================================================================

# Input umur dan status kepemilikan SIM
umur = int(input("Masukkan umur Anda: "))
memiliki_sim_input = input("Apakah Anda memiliki SIM? (true/false): ").strip().lower()

# Konversi input string ke tipe booelan
punya_sim = memiliki_sim_input == "true"

# Evaluasi kelayakan mengemudi (menggunakan tanda kurung eksplisit untuk precedence)
boleh_mengemudi = (umur >= 17) and (punya_sim == True)

print(f"\nHasil Evaluasi Kelayakan Mengemudi:")
print(f"Umur >= 17: {umur >= 17}")
print(f"punya_sim == True: {punya_sim == True}")
print(f"Kelayakan Mengemudi: {boleh_mengemudi}")
```

```python
# Jawaban Latihan B.2 No 4
# ============================================================================
# Nama       : Rehan
# NIM        : 4232501051
# Tanggal    : 22 September 2026
# Tujuan     : Menghitung efesiensi termal pembangkit listrik berdasarkan energi
#              input dan output (mj) menggunakan operator aritmatika &  f-string.
# ============================================================================
energi_input = float(input("Masukkan Energi Input (MJ) : "))
energi_output = float(input("Masukkan Energi Output (MJ): "))

efesiensi = energi_output / energi_input * 100
print(f"\nEfesiensi termal pembangkit adalah sebesar {efesiensi:.2f}%")
```
